```python
# Task 1: Netflix Data Cleaning & Preparation
# Consolidated cleaning steps based on inspection findings above :

# 1. Replace 'Not Given' placeholder text with real NaN in director and country columns.
#    Turns 'Not Given' text into a true blank, so summaries and counts automatically
#    skip these rows instead of accidentally counting 'Not Given' as if it were a real answer.
df['director'] = df['director'].replace('Not Given', pd.NA)
df['country'] = df['country'].replace('Not Given', pd.NA)

# 2. Convert date_added from text to an actual datetime type.
#    Enables chronological sorting, filtering, and year/month extraction later.
df['date_added'] = pd.to_datetime(df['date_added'], format='%m/%d/%Y')

# 3. Remove leading/trailing spaces from titles.
df['title'] = df['title'].str.strip()

# 4. Standardize rating labels: UR (Unrated) and NR (Not Rated) mean the same thing.
#    Folding into NR since it has far more rows (79 vs 3).
df['rating'] = df['rating'].replace('UR', 'NR')

# 5. 'Soviet Union' and 'West Germany' kept as-it-is: only 2 rows affected, and remapping to
#    modern country names (e.g. Russia, Germany) would misrepresent historical fact. No code needed.

# 6. Remove duplicate records.
#    "A duplicate = identical in every column EXCEPT show_id.This covers the 3 date-formatted like title pairs AND the 'Consequences' pair".
cols_without_id = df.columns.drop('show_id').tolist()

#  Audit trail: view exactly which rows are about to be removed
removed_rows = df[df.duplicated(subset=cols_without_id, keep='first')]
print(f"Rows to be removed: {len(removed_rows)}") # expect 4
print(removed_rows[['show_id', 'title', 'type']])

df= df.drop_duplicates(subset=cols_without_id, keep='first')

# 7. Reset the index after dropping rows, so the DataFrame has clean, sequential row numbers.
df = df.reset_index(drop=True)  # (drop=True)- discards the old index

# Sanity check: confirm final row count and remaining data-quality counts.
print(f"Final row count: {len(df)}")            # expect 8790 - 4 = 8786
print(df['director'].isna().sum(), "rows with missing director (now proper NaN)")   # expect 2587 and not 2588(as one duplicate pair had no directors)
print(df['country'].isna().sum(), "rows with missing country (now proper NaN)")     # expect 287
print(df['rating'].value_counts().get('UR', 0), "rows still labelled UR")           # expect 0
print(df[df['title'] != df['title'].str.strip()].shape[0], "titles with stray spaces")  # expect 0
```

```python
# Post-cleaning verification 
# assert <condition>, "message": silent if True; stops the notebook if False
df.info()   # confirm column dtypes after cleaning

# show_id must still be unique
assert df['show_id'].is_unique, "Duplicate show_id values found"

# No duplicate records, ignoring show_id
assert df.duplicated(subset=cols_without_id).sum() == 0, "Duplicate records remain (ignoring show_id)"

# date_added is a real datetime column with no missing values
#    (is_datetime64_any_dtype is used because the exact dtype varies by pandas version)
assert pd.api.types.is_datetime64_any_dtype(df['date_added']), "date_added is not a datetime column"
assert df['date_added'].notna().all(), "date_added contains missing values"

# release_year in a plausible range
assert df['release_year'].between(1900, 2021).all(), "release_year outside 1900-2021"

print("ALL CHECKS PASSED")
```
