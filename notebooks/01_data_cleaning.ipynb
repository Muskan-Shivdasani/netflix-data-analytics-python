```python
# --- Task 1: Netflix Data Cleaning & Preparation ---
# Consolidated cleaning steps based on inspection findings above.

# 1. Replace 'Not Given' placeholder text with real NaN in director and country columns.
#    Turns 'Not Given' text into a true blank, so summaries and counts automatically
#    skip these rows instead of accidentally counting 'Not Given' as if it were a real answer.
df['director'] = df['director'].replace('Not Given', pd.NA)
df['country'] = df['country'].replace('Not Given', pd.NA)

# 2. Convert date_added from text to an actual datetime type.
#    Enables chronological sorting, filtering, and year/month extraction later.
df['date_added'] = pd.to_datetime(df['date_added'])

# 3. Remove leading/trailing spaces from titles.
#    One title ("Consequences", show_id s6530) had a hidden trailing space.
df['title'] = df['title'].str.strip()

# 4. Standardize rating labels: UR (Unrated) and NR (Not Rated) mean the same thing.
#    Folding into NR since it has far more rows (79 vs 3).
df['rating'] = df['rating'].replace('UR', 'NR')

# 5. 'Soviet Union' and 'West Germany' kept as-is: only 2 rows affected, and remapping to
#    modern country names (e.g. Russia, Germany) would misrepresent historical fact. No code needed.

# 6. Remove the 6 rows identified as duplicate + corrupted-title records.
#    These 3 pairs share identical data across all columns except show_id, and their
#    'title' field was corrupted into date-like text (e.g. '15-Aug') prior to receipt —
#    the original titles are unrecoverable, and a title-less record has no analytical value.
corrupted_ids = ['s3963', 's5967', 's4523', 's5966', 's3997', 's5965'] # show_ids of those records
df = df[~df['show_id'].isin(corrupted_ids)]

# 7. Reset the index after dropping rows, so the DataFrame has clean, sequential row numbers.
df = df.reset_index(drop=True)

# Sanity check: confirm final row count and remaining data-quality counts.
print(f"Final row count: {len(df)}")            # expect 8790 - 6 = 8784
print(df['director'].isna().sum(), "rows with missing director (now proper NaN)")   # expect 2586
print(df['country'].isna().sum(), "rows with missing country (now proper NaN)")     # expect 287
print(df['rating'].value_counts().get('UR', 0), "rows still labelled UR")           # expect 0
print(df[df['title'] != df['title'].str.strip()].shape[0], "titles with stray spaces")  # expect 0
```
